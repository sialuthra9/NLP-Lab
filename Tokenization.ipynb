```python
#word tokenizer
text ="Ram, Shayam, and I are good friends. We both Play cricket."
tokens = word_tokenize(text)
print(tokens)
```

```python
#sentence tokenizer
text ="Ram, Shayam, and I are good friends. We both Play cricket."
tokens = sent_tokenize(text)
print(tokens)
```

```python
#removing 
import string
text=" ! hey , i am %$*< ,"

tokens= word_tokenize(text)
new_text=[]
for i in tokens:
    if i not in string.punctuation:
     new_text.append(i)
new_text
```

```python
#stop word removal
from nltk.corpus import stopwords 
nltk.download('stopwords')

text ="Ram, Shayam, and I are good friends. We both Play cricket." 
tokens = word_tokenize(text) 
clean_text=[] 
for i in tokens: 
    if i.lower() not in stopwords.words('english'):
      clean_text.append(i) 
    
clean_text
```

```python
#stemming
from nltk.stem import PorterStemmer
stemming=PorterStemmer()

words=["Playing","Plays","played","unplayed","players","Playful"]
for i in words:
    print(i,"->",stemming.stem(i.lower()))
```

```python
#lemmatization(Slow)
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemm=WordNetLemmatizer()
words=["fairly","easily","happy","happiness","unplayed","playeed"]
for i in words:
    print(lemm.lemmatize(i))
    
```
